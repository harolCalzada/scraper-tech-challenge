## Importar las dependencias necesarias

```python
import sys
sys.path.append('.')

from src.infrastructure.adapters.mercadolibre_scraper import MercadoLibreScraper
from src.infrastructure.adapters.excel_repository import ExcelRepository
from src.domain.services.product_matcher import ProductMatcher
from src.application.use_cases.scrape_products import ScrapeProductsUseCase
```

## Configurar los componentes

```python
# Configurar rutas de archivos
input_file = 'data/input.xlsx'  # Los archivos se guardarán en el directorio data
output_file = 'data/output.xlsx'

# Crear directorio data si no existe
!mkdir -p data

# Inicializar componentes
scraper = MercadoLibreScraper()
repository = ExcelRepository()
matcher = ProductMatcher()

# Crear el caso de uso
use_case = ScrapeProductsUseCase(scraper, repository, matcher)
```

## Subir archivo de entrada

Sube el archivo Excel con los productos a buscar:

```python
from google.colab import files
uploaded = files.upload()

# Guardar el archivo subido en el directorio data
with open('data/input.xlsx', 'wb') as f:
    f.write(uploaded[next(iter(uploaded))])
```

## Ejecutar el scraper

```python
import asyncio

async def run_scraper():
    results = await use_case.execute(input_file, output_file)
    return results

# Ejecutar el scraper
results = await run_scraper()

print(f"Found {len(results)} matching products")
print(f"Results saved to: {output_file}")
```
