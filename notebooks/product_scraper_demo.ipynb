# Product Scraper Demo

This notebook demonstrates the basic functionality of the product scraper.

```python
import sys
import asyncio
from pathlib import Path

# Add project root to Python path
project_root = Path.cwd().parent
sys.path.append(str(project_root))

from src.infrastructure.adapters.excel_repository import ExcelRepository
from src.infrastructure.adapters.mock_scraper import MockScraper
from src.application.use_cases.search_products_use_case import SearchProductsUseCase
```

## Setup Components

```python
# Initialize components
repository = ExcelRepository()
scraper = MockScraper()
use_case = SearchProductsUseCase(scraper, repository)
```

## Process Input File

```python
# File paths
input_file = '../input/products.xlsx'
output_file = '../output/results.xlsx'

# Create directories if they don't exist
Path('../input').mkdir(exist_ok=True)
Path('../output').mkdir(exist_ok=True)

# Run the use case
async def run_scraper():
    results = await use_case.execute(input_file, output_file)
    return results

# Execute the scraper
results = await run_scraper()

print(f"Found {len(results)} matching products")
print(f"Results saved to: {output_file}")
```
